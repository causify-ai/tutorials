# Real-Time Bitcoin Sentiment Analysis Using txtai

This notebook shows how to build a simple AI-powered search engine using `txtai`.

We'll use a few example Bitcoin-related headlines to show how to use semantic search to find them based on meaning.

This is a small demo to help you understand how txtai works.

## Step 1: Import our helper class

We created a Python file `txtai_sentiment_search.py` with a class that helps us:
- Build a txtai index
- Run a search query

```python
from txtai.embeddings import Embeddings

class TxtaiSentimentSearch:
    def __init__(self, model: str = "sentence-transformers/all-MiniLM-L6-v2"):
        """
        Initializes the TxtaiSentimentSearch class with a specified sentence transformer model.

        :param model: Name of the Hugging Face model to use for semantic embeddings
        """
        # initialize the txtai Embeddings index using the specified transformer model
        self.index = Embeddings({"path": model})

        # create list to store original sentences for retrieval after search
        self.sentences = []

    def build_index(self, data):
        """
        Builds a semantic search index from a list of text data.

        :param data: List of textual sentences or headlines to index
        """
        # Save original text data for retrieval
        self.sentences = data

        # Format each sentence as a dict with an id and text, as expected by txtai
        documents = [{"id": i, "text": text} for i, text in enumerate(data)]

        # Index the documents using txtai
        self.index.index(documents)

    def search(self, query, top_k=1):
        """
        Searches the txtai index for the most semantically similar results.

        :param query: A natural language query string
        :param top_k: The number of top results to return
        :return: List of matched sentences (strings)
        """
        # Run semantic search; returns list of tuples (id, score)
        results = self.index.search(query, top_k)

        # Use the ids to retrieve original text from self.sentences
        return [self.sentences[i] for i, _ in results]
```

## Step 2: Sample Bitcoin Headlines

This is just a list of short example headlines. In the full project, this data will come from real-time news APIs.

```python
headlines = [
    "Bitcoin price hits new all-time high!",
    "Huge drop in crypto markets today.",
    "Investors remain confident in Bitcoin.",
    "SEC announces new crypto rules.",
    "Big Bitcoin selloff causes panic."
]
```

## Step 3: Build the semantic index using txtai

Here we use the helper class to create an index of the headlines.

Behind the scenes, txtai turns each sentence into a vector (embedding), which makes it easier to search by meaning.

```python
# Create the search engine
search_engine = TxtaiSentimentSearch()

# Build the index with our list of headlines
search_engine.build_index(headlines)
```

## Step 4: Ask a question and search the headlines

Now we try a question like: "Why is Bitcoin falling?"

txtai compares this question to all indexed headlines and finds the closest matches.

```python
query = "Why is Bitcoin falling?"

# Search using the query
results = search_engine.search(query)

# Print results
for sentence in results:
    print(f"- {sentence}")
```
