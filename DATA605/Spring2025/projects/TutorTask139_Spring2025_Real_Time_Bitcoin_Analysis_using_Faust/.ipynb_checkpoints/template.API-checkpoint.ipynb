````python
import requests
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime
import time
import statistics

# Set up the plot style
plt.style.use('fivethirtyeight')
````

````python
def fetch_bitcoin_data():
    """Fetch current Bitcoin data from CoinGecko with additional metrics."""
    url = 'https://api.coingecko.com/api/v3/coins/bitcoin'
    params = {'localization': 'false', 'tickers': 'false', 'community_data': 'false', 'developer_data': 'false'}
    
    try:
        response = requests.get(url, params=params)
        if response.status_code == 200:
            data = response.json()
            market_data = data['market_data']
            return {
                'timestamp': datetime.utcnow(),
                'price': market_data['current_price']['usd'],
                'volume_24h': market_data['total_volume']['usd'],
                'price_change_24h': market_data['price_change_percentage_24h']
            }
        else:
            print(f"Error fetching data: {response.status_code} {response.reason}")
            return None
    except Exception as e:
        print(f"Error: {str(e)}")
        return None
````

````python
# Initialize data storage
timestamps = []
prices = []
volumes = []
price_changes = []

# Create figure for real-time plotting
plt.figure(figsize=(15, 10))

try:
    while True:
        # Fetch new data
        data = fetch_bitcoin_data()
        
        if data:
            # Append new data
            timestamps.append(data['timestamp'])
            prices.append(data['price'])
            volumes.append(data['volume_24h'])
            price_changes.append(data['price_change_24h'])
            
            # Calculate analytics if we have enough data
            if len(prices) >= 2:
                # Calculate moving averages
                ma_5 = statistics.mean(prices[-5:]) if len(prices) >= 5 else prices[-1]
                ma_15 = statistics.mean(prices[-15:]) if len(prices) >= 15 else prices[-1]
                ma_30 = statistics.mean(prices[-30:]) if len(prices) >= 30 else prices[-1]
                
                # Calculate volatility
                volatility = statistics.stdev(prices[-30:]) if len(prices) >= 30 else 0
                
                # Calculate price change
                price_change = ((prices[-1] - prices[-2]) / prices[-2]) * 100
                
                # Clear previous plots
                plt.clf()
                
                # Create subplot for price and MAs
                plt.subplot(2, 1, 1)
                plt.plot(timestamps, prices, 'b-', label='Price', linewidth=2)
                if len(prices) >= 5:
                    plt.plot(timestamps[-5:], [ma_5]*5, 'r--', label='5-point MA')
                if len(prices) >= 15:
                    plt.plot(timestamps[-15:], [ma_15]*15, 'g--', label='15-point MA')
                if len(prices) >= 30:
                    plt.plot(timestamps[-30:], [ma_30]*30, 'y--', label='30-point MA')
                
                plt.title('Bitcoin Price Analysis')
                plt.xlabel('Time')
                plt.ylabel('Price (USD)')
                plt.legend()
                plt.grid(True)
                
                # Create subplot for volume
                plt.subplot(2, 1, 2)
                plt.bar(timestamps, volumes, alpha=0.7, color='blue')
                plt.title('Trading Volume (24h)')
                plt.xlabel('Time')
                plt.ylabel('Volume (USD)')
                plt.grid(True)
                
                # Adjust layout
                plt.tight_layout()
                
                # Print analysis
                print(f"\nAnalysis ({datetime.now().strftime('%Y-%m-%d %H:%M:%S')})")
                print(f"Current Price: ${prices[-1]:,.2f}")
                print(f"5-point MA: ${ma_5:,.2f}")
                print(f"15-point MA: ${ma_15:,.2f}")
                print(f"30-point MA: ${ma_30:,.2f}")
                print(f"Volatility: {volatility:,.2f}")
                
                if abs(price_change) > 2:
                    print(f"[Alert] Significant price change: {price_change:.2f}%")
                
                # Show plot
                plt.draw()
                plt.pause(0.1)
            
            # Wait for 60 seconds before next update
            time.sleep(60)
            
except KeyboardInterrupt:
    print("\nStopping price tracking...")
````
