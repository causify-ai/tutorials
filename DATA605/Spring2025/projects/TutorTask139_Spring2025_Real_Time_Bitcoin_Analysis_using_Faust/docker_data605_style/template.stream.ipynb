```python
import faust
from datetime import datetime

# Initialize Faust app
app = faust.App(
    'btc-price-stream',
    broker='kafka://localhost:9092',
    value_serializer='json'
)
```

```python
# Define the schema for Bitcoin price data
class BitcoinPrice(faust.Record, serializer='json'):
    timestamp: str
    price: float
```

```python
# Define Kafka topic for price data
price_topic = app.topic('btc_price_topic', value_type=BitcoinPrice)

# Agent for processing the data stream
@app.agent(price_topic)
async def process_price(prices):
    async for price in prices:
        print(f"[{price.timestamp}] Received Bitcoin Price: ${price.price}")
        # Extend with additional logic (e.g., filtering, rolling average, etc.)
```
