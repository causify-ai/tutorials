````python
# A simple function for demo/testing purposes
def analyze_bitcoin(price_list):
    avg_price = sum(price_list) / len(price_list)
    logging.info("Calculated average price: $%.2f", avg_price)
    return avg_price
````

````python
# Simulate BTC prices and analyze them
btc_prices = [64000, 64500, 65000, 63800, 64200]
average_price = analyze_bitcoin(btc_prices)
print("📊 Average Bitcoin Price:", average_price)
````
