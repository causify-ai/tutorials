# Example: Bitcoin Trade Processing

This notebook shows a simple example of how to use the Bitcoin Trade Processing API.

- We simulate a Bitcoin trade manually.
- We process it using the API functions.
- Monitoring, aggregation, anomaly detection, sentiment analysis, and Slack alerts happen automatically.


```python
# Import the main API function
from btc_trade_API import start_monitoring, process_trade
```

```python
# Start monitoring server
start_monitoring(port=8000)
```

```python
# Create a dummy Bitcoin trade for testing
example_trade = {
    "price": 50234.50,
    "time": "2025-04-26T19:00:00Z"
}
```

```python
# Process the dummy trade
process_trade(example_trade)
```

## Notes:

- The trade gets asynchronously aggregated into OHLC metrics.
- Anomalous trades (3-sigma deviation) will trigger Slack alerts.
- Latest Bitcoin subreddit sentiment is fetched and logged.
- Prometheus metrics are available at [http://localhost:8000/metrics](http://localhost:8000/metrics).

✅ This completes the example usage of the Bitcoin Trade Processing System.
