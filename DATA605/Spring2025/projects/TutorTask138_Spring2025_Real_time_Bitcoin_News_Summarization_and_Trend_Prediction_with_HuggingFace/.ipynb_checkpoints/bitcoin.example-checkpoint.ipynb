````python
import pandas as pd
from bitcoin_utils import (
    fetch_bitcoin_news_from_newsapi,
    summarize_and_analyze_articles
)


from dotenv import load_dotenv
import os
load_dotenv()
API_KEY = os.getenv("NEWSAPI_KEY")
````

````python
from bitcoin_utils import fetch_bitcoin_news_multiple_pages

# Fetch 100 articles per query using different variations
articles_1 = fetch_bitcoin_news_from_newsapi(API_KEY, query="bitcoin", page_size=100)
articles_2 = fetch_bitcoin_news_from_newsapi(API_KEY, query="bitcoin price", page_size=100)


# Combine all articles
articles = articles_1 + articles_2 

print(f"Fetched a total of {len(articles)} articles.")

print(f"Fetched {len(articles)} articles.")

````

````python
articles
````

````python
results = summarize_and_analyze_articles(articles)
````

````python
df = pd.DataFrame(results)
df.to_csv("summarized_sentiment_articles1.csv", index=False)
print("Saved results to CSV.")
````

````python

df = pd.read_csv("summarized_sentiment_articles1.csv")
df.head()
````

````python
import matplotlib.pyplot as plt
df['sentiment'].value_counts().plot(kind='bar', title='Sentiment Distribution')
plt.xlabel("Sentiment")
plt.ylabel("Article Count")
plt.grid(True)
plt.tight_layout()
plt.show()
````

````python
sentiment_map = {
    "POSITIVE": 1,
    "NEGATIVE": -1,
    }
df['sentiment_score'] = df['sentiment'].map(sentiment_map)
````

````python
print(df.columns)
````

````python
df['date'] = pd.to_datetime(df['publishedAt']).dt.date
````

````python
daily_sentiment = df.groupby("date")["sentiment_score"].mean().reset_index()
daily_sentiment.columns = ["date", "avg_sentiment"]
````
