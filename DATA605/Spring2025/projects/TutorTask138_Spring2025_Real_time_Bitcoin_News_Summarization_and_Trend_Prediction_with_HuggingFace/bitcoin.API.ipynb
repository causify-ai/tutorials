# Summarization

#  Sentiment Analysis


## Analyze Sentiment of the First News Article


```python
# Analyze the sentiment of the first Bitcoin news article using analyze_sentiment().
sentiment = analyze_sentiment(news_articles[0])

# Display the sentiment analysis result.
print("\nSentiment of the Article:")
print(sentiment)
```
