```python
%matplotlib notebook
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import pandas as pd
import os
from producer.btc_producer import run_producer
from consumer.btc_consumer import run_consumer
import multiprocessing
import time
```

```python
producer_process = multiprocessing.Process(target=run_producer)
consumer_process = multiprocessing.Process(target=run_consumer)

producer_process.start()
consumer_process.start()

# Allow data to build up
time.sleep(5)
```

```python
fig, ax = plt.subplots()
x_data, y_data = [], []
line, = ax.plot([], [], lw=2)
ax.set_title("Real-Time Bitcoin Price")
ax.set_xlabel("Time")
ax.set_ylabel("Price (USD)")

csv_file =  r'C:\Users\amans\OneDrive\Desktop\tutorials\DATA605\Spring2025\projects\TutorTask193_Spring2025_Real_Time_Bitcoin_Data_Ingestion_and_Analysis_using_PyKafka\bitcoin-streaming-analysis\project\data\btc_data.csv'

def update(frame):
    if os.path.exists(csv_file):
        df = pd.read_csv(csv_file)
        x_data = pd.to_datetime(df['timestamp'], unit='s')
        y_data = df['price']
        line.set_data(x_data, y_data)
        ax.relim()
        ax.autoscale_view()
    return line,

ani = animation.FuncAnimation(fig, update, interval=2000)
plt.show()
```

```python
# Optional: Stop processes
producer_process.terminate()
consumer_process.terminate()
```
