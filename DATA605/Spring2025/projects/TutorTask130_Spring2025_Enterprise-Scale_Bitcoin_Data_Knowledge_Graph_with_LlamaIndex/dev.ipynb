### Raw Block fields

'hash': Unique identifier for the block <br>
'confirmations': Number of confirmations (indicates how deep in the blockchain) <br>
'height': Block number/position in the blockchain <br>
'version': Bitcoin protocol version used <br>
'merkleroot': Root of the merkle tree of all transactions in the block <br>
'time': Block timestamp <br>
'nTx': Number of transactions in the block <br>
'tx': List of transaction IDs contained in the block <br>
'previousblockhash'/'nextblockhash': Links to previous/next blocks <br>
'difficulty': Mining difficulty at time of block creation <br>
