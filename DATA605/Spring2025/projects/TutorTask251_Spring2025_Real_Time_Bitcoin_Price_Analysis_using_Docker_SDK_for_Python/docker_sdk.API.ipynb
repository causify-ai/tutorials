# Real-Time Bitcoin Price Analysis using Docker SDK for Python

This notebook demonstrates how to use the Docker SDK for Python to automate container orchestration in a real-time Bitcoin analytics pipeline. It includes:

- Automated building and launching of Docker containers for InfluxDB, Grafana, and BTC fetcher.
- Continuous ingestion of real-time and historical BTC/USD price data from the CryptoCompare API, tagged with `"source"`.
- Visualization of both real-time and historical data in Grafana with source-tag filtering (`realtime` vs `historical`).

## Imports

```python
import logging
import time
import docker_sdk_utils as utils
```

## 1. List Docker Images

This demonstrates how to use the wrapper function `utils.list_docker_images()` to list all Docker images installed locally.  

 This is useful to check that the required Docker images (BTC fetcher, InfluxDB, Grafana) are available before launching the containers.

```python
utils.run_pipeline()

images = utils.list_docker_images()
print([img.tags for img in images])
```

## 2. List Docker Containers

Here we use `utils.list_docker_containers(all=True)` to list all active and stopped containers.  
This helps verify which containers exist, especially useful for cleanup or inspection.

```python
containers = utils.list_docker_containers()
for container in containers:
    print(container.name, container.status)
```

## 3. Pull Docker Image

We can programmatically pull any Docker image using `utils.pull_docker_image()`.  
This avoids manual CLI steps and ensures required images are present for later use.

```python
utils.pull_docker_image('influxdb:latest')
```

## 4. Start InfluxDB Container

The `utils.start_influxdb_container()` function creates and starts an InfluxDB container.  
This can be used as a lightweight local time-series DB for any data pipeline.

```python
test_container = utils.start_influxdb_container(container_name='influxdb_api_demo', port=8090)
print(f"Started container: {test_container.name}")
```

## 5. Stop a Container

This function call demonstrates how to stop and remove a Docker container using `utils.stop_docker_container()`.  
Always stop unused containers to keep the system clean.

```python
import time
time.sleep(5)
utils.stop_docker_container('influxdb_api_demo')
print("Stopped and removed container.")
```
