```python
import pandas as pd

# Step 1: Load all three CSVs
df1 = pd.read_csv("bitcoin_real_time_data.csv")
df2 = pd.read_csv("bitcoin_real_time_data1.csv")
df3 = pd.read_csv("bitcoin_real_time_data2.csv")

# Step 2: Combine them together
df_combined = pd.concat([df1, df2, df3], ignore_index=True)

# Step 3: Optional - Sort by timestamp if needed
df_combined['timestamp'] = pd.to_datetime(df_combined['timestamp'], errors='coerce')
df_combined = df_combined.sort_values(by='timestamp')

# Step 4: Reset index and add new S.No again
df_combined = df_combined.reset_index(drop=True)
df_combined.insert(0, 'S.No', range(1, len(df_combined) + 1))

# Step 5: Save the combined data
df_combined.to_csv("bitcoin_combined_data.csv", index=False)

# Step 6: Download the combined file (if in Google Colab)
from google.colab import files
files.download("bitcoin_combined_data.csv")
```

```python
import pandas as pd
from dataprep.clean import clean_date
```

```python
df = pd.read_csv('bitcoin_combined_data.csv')
```

```python
df['price_usd'] = df['price_usd'].fillna(method='ffill')
```

```python
df['timestamp'] = pd.to_datetime(df['timestamp'], errors='coerce')
```

```python
df['price_usd'] = pd.to_numeric(df['price_usd'], errors='coerce')
```

```python
df['price_usd'] = df['price_usd'].fillna(method='ffill')
```

```python
df['timestamp'].head()
```

```python
# First, show all rows and columns without limit
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)

# Now print the entire DataFrame
print(df)
```

**EDA**

```python
# Step 1: Quick Overview
print(df.info())
print(df.describe())

# Step 2: Check missing values
print(df.isnull().sum())

# Step 3: Plot Bitcoin price trend
import matplotlib.pyplot as plt
plt.figure(figsize=(15,6))
plt.plot(df['timestamp'], df['price_usd'])
plt.title('Bitcoin Price Trend Over Time')
plt.xlabel('Timestamp')
plt.ylabel('Price (USD)')
plt.xticks(rotation=45)
plt.grid()
plt.show()

# Step 4: Correlation
import seaborn as sns
plt.figure(figsize=(6,5))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()


```

*** Bitcoin Price - Moving Average Plot (Trend Smoother)***


```python
df['MA_5'] = df['price_usd'].rolling(window=5).mean()
df['MA_10'] = df['price_usd'].rolling(window=10).mean()

plt.figure(figsize=(15,6))
plt.plot(df['timestamp'], df['price_usd'], label='Price (USD)')
plt.plot(df['timestamp'], df['MA_5'], label='5-Point MA')
plt.plot(df['timestamp'], df['MA_10'], label='10-Point MA')
plt.title('Bitcoin Price with Moving Averages')
plt.xlabel('Timestamp')
plt.ylabel('Price (USD)')
plt.xticks(rotation=45)
plt.legend()
plt.grid()
plt.show()
```

**Bitcoin** **Price** **Distribution**

```python
plt.figure(figsize=(10,5))
sns.histplot(df['price_usd'], bins=30, kde=True)
plt.title('Distribution of Bitcoin Prices')
plt.xlabel('Price (USD)')
plt.ylabel('Frequency')
plt.grid()
plt.show()
```

**Bitcoin Price Change Over Time**

```python
df['price_change'] = df['price_usd'].diff()

plt.figure(figsize=(15,6))
plt.plot(df['timestamp'], df['price_change'])
plt.title('Bitcoin Price Change Over Time')
plt.xlabel('Timestamp')
plt.ylabel('Price Change (USD)')
plt.xticks(rotation=45)
plt.grid()
plt.show()
```

**Missing Value Heatmap**

```python
plt.figure(figsize=(8,4))
sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
plt.title('Missing Values Heatmap')
plt.show()
```

**Resample Bitcoin Price by Time (Hourly Average)**


```python
# First, make sure timestamp is datetime
df['timestamp'] = pd.to_datetime(df['timestamp'])

# Set timestamp as index
df.set_index('timestamp', inplace=True)

# Resample: Hourly mean price
df_hourly = df['price_usd'].resample('H').mean()

# Plot hourly average
plt.figure(figsize=(15,6))
plt.plot(df_hourly)
plt.title('Hourly Average Bitcoin Price')
plt.xlabel('Hour')
plt.ylabel('Price (USD)')
plt.grid()
plt.show()

# Reset index if needed after this
df.reset_index(inplace=True)
```

**Outlier Detection (Boxplot)**

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df['price_usd'])
plt.title('Boxplot of Bitcoin Prices')
plt.grid()
plt.show()
```

**Price Change vs Price Plot**

```python
plt.figure(figsize=(10,6))
plt.scatter(df['price_usd'], df['price_change'], alpha=0.6)
plt.title('Price vs Price Change')
plt.xlabel('Price (USD)')
plt.ylabel('Price Change (USD)')
plt.grid()
plt.show()
```

**Rolling Standard Deviation**

```python
# Calculate rolling std dev
df['rolling_std'] = df['price_usd'].rolling(window=10).std()

# Plot
plt.figure(figsize=(15,6))
plt.plot(df['timestamp'], df['rolling_std'])
plt.title('Rolling Standard Deviation (Volatility) of Bitcoin Prices')
plt.xlabel('Timestamp')
plt.ylabel('Rolling Std Dev')
plt.grid()
plt.show()
```
