### Real-Time Bitcoin Price Streaming: Sample API Usage

```python
import warnings
warnings.filterwarnings('ignore')

from bitcoin_utils import BitcoinPriceSource
import itertools

# Initialize source with window
source = BitcoinPriceSource()

# Simulate just one fetch loop iteration for demonstration
for i, (timestamp, price) in zip(range(10), source):
    print(f"[{i+1}] Timestamp: {timestamp}, Price: ${price:.2f}")
```

The output shows Bitcoin price fetches every 30 seconds along with computed statistics once enough data points are collected## Price Fetch Logs
Each fetch prints the current UTC timestamp and Bitcoin price. It also prints a count of the fetch number, the timestamp in milliseconds, and the price.

## Computed Metrics
After collecting 10 price points (the window size), the script calculates and displays:

- **Moving Average (MA)** of prices in the window
- **Standard Deviation (StdDev)**, measuring price volatility
- **Exponential Moving Average (EMA)**, which weights recent prices more
- **Maximum and Minimum** prices in the window
- **Trend indicator**: -1 means the price is trending down, 1 means up
- **Cumulative Return**: percentage price change over the window
- **24h Change**: price change in last 24 hours reported by the API

## Summary
The script fetches prices repeatedly, and after every 10 samples, it summarizes recent price behavior to help analyze short-term market trends and volatility.
