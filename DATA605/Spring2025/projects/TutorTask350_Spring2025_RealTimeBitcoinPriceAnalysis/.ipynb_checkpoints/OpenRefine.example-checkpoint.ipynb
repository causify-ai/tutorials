````python
# Fetch 7 days of 15min interval data
btc_df = fetch_bitcoin_data_kucoin(days=7, interval='15min')

# Show sample data
btc_df.head()
````

````python
from bitcoin_utils import load_cleaned_data, validate_cleaned_data

# Load cleaned data
cleaned_df = load_cleaned_data('bitcoin_price_analysis_using_OpenRefine.csv')

# Run validation checks
validation_passed = validate_cleaned_data(cleaned_df)

if validation_passed:
    print("\n=== VALIDATION SUCCESS ===")
    print("Data is clean and ready for analysis!")
    display(cleaned_df.head(3))
else:
    print("\n=== VALIDATION FAILED ===")
    print("Fix issues in OpenRefine before proceeding")
````

````python
from bitcoin_utils import resample_data
hourly_df = resample_data(cleaned_df, '1H')
hourly_df.head()
````

````python
# In notebook
import matplotlib.pyplot as plt

# Plot price trends
plt.figure(figsize=(14, 6))
plt.plot(hourly_df['close'], label='Hourly Close Price')
plt.title('Bitcoin Price Trends')
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.legend()
plt.show()

# Volume analysis
plt.figure(figsize=(14, 6))
plt.bar(hourly_df.index, hourly_df['volume'], color='orange')
plt.title('Trading Volume')
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.show()
````

````python
from bitcoin_utils import calculate_technical_indicators

analyzed_df = calculate_technical_indicators(cleaned_df)
print("\nTechnical Indicators Added for Analysis:")
display(analyzed_df[['timestamp', 'ma_7', 'ma_24', 'intraday_volatility', 'daily_momentum']].head())
````

````python
plt.figure(figsize=(20,10))

# Price and Moving Averages
plt.subplot(2, 2, 1)
plt.plot(analyzed_df['timestamp'], analyzed_df['close'], label='Closing Price', color='blue')
plt.plot(analyzed_df['timestamp'], analyzed_df['ma_7'], label='7-Period MA', linestyle='--', color='orange')
plt.plot(analyzed_df['timestamp'], analyzed_df['ma_24'], label='24-Period MA', color='green')
plt.fill_between(analyzed_df['timestamp'], analyzed_df['upper_band'], analyzed_df['lower_band'], alpha=0.2, color='gray')

plt.title('Price Trend with Moving Averages & Bollinger Bands')
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.legend()
````
