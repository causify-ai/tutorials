**Real-Time Bitcoin Price Streaming with CoinGecko API & PyArrow**


This notebook demonstrates how to stream live Bitcoin price data from the CoinGecko API and structure it efficiently using PyArrow. It fetches price data at regular intervals, stores it in memory, and writes it to a Parquet file for downstream analytics or storage.

**Objectives**


Fetch real-time Bitcoin price data.

Handle CoinGecko API response and rate limits.

Store data in-memory using PyArrow tables.

Write data to disk in Parquet format.

**Loading Libraries and Environment Variables**

```python
from dotenv import load_dotenv
import os
import pyarrow.parquet as pq
from utils import stream_prices
```

```python
# Load API key
load_dotenv("vals.env")
api_key = os.getenv("API_KEY")
```

**Streaming Data and writing as Parquet**

```python
# Stream live price data for Bitcoin
arrow_table = stream_prices(api_key, duration_sec=300, interval_sec=30)

# Optional: Save to Parquet
pq.write_table(arrow_table, "datalake/bitcoin_price_stream.parquet")
print("Data saved to datalake/bitcoin_price_stream.parquet")
```
