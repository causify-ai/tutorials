```python
import pandas as pd
from datetime import datetime, timedelta, UTC

# Load data
df = pd.read_csv("data/bitcoin_prices.csv")

# Convert timestamps to UTC-aware datetime
df['timestamp'] = pd.to_datetime(df['timestamp']).dt.tz_localize('UTC')

# Get current time and filter last 30 minutes
now = datetime.now(UTC)
start_time = now - timedelta(minutes=30)
df_30min = df[df['timestamp'] >= start_time].sort_values('timestamp')

# Format the timestamp nicely for display
df_30min['Day'] = df_30min['timestamp'].dt.strftime("%A")
df_30min['Date'] = df_30min['timestamp'].dt.strftime("%B %d, %Y")
df_30min['Time (UTC)'] = df_30min['timestamp'].dt.strftime("%I:%M %p")
df_30min['Price (USD)'] = df_30min['price'].map("${:,.2f}".format)

# Select and reorder columns for display
display_table = df_30min[['Day', 'Date', 'Time (UTC)', 'Price (USD)']]

# Display as a styled table
display(display_table.style.set_caption("📊 Bitcoin Prices in the Last 30 Minutes").set_table_styles([{
    'selector': 'caption',
    'props': [('font-size', '16px'), ('font-weight', 'bold')]
}]))
```

```python
# Record BTC prices every 10 seconds (12 times = 2 minutes)

print(" Recording live prices for 120 seconds...\n")
for i in range(12):  # 12 records x 10 seconds = 2 minutes
    record_price()
    print(f" Recorded price {i+1}/12")
    time.sleep(10)

print("\n Finished collecting data.")
```

```python
import pandas as pd
from datetime import datetime
from IPython.display import display

# Load all recorded data
df = pd.read_csv("data/bitcoin_prices.csv")

# Parse timestamps
df['timestamp'] = pd.to_datetime(df['timestamp'])

# Get the last 12 entries only (1 per record attempt)
df_last12 = df.tail(12).copy()

# Format timestamp for better readability
df_last12['Day'] = df_last12['timestamp'].dt.strftime("%A")
df_last12['Date'] = df_last12['timestamp'].dt.strftime("%B %d, %Y")
df_last12['Time (UTC)'] = df_last12['timestamp'].dt.strftime("%I:%M:%S %p")

# Format price (and flag any failed fetch as NaN or missing)
df_last12['Bitcoin Price (USD)'] = df_last12['price'].apply(
    lambda p: "${:,.2f}".format(p) if pd.notnull(p) else "❌ Error"
)

# Final display table
display_table = df_last12[['Day', 'Date', 'Time (UTC)', 'Bitcoin Price (USD)']]

# Display it as a clean styled table
display(
    display_table.style.set_caption("📊 Last 12 Bitcoin Price Fetches (from 2-minute loop)").set_table_styles([{
        'selector': 'caption',
        'props': [('font-size', '16px'), ('font-weight', 'bold')]
    }])
)
```

```python
###  Cell 9: Bonus — Price Diff + Rolling Avg Plot

if not df_clean.empty:
    plt.figure(figsize=(10, 5))
    plt.plot(df_clean['timestamp'], df_clean['price_diff'], label='Price Δ (USD)', color='orange')
    plt.plot(df_clean['timestamp'], df_clean['rolling_avg'], label='Rolling Avg', color='green')
    plt.title("Bitcoin Price Δ and Rolling Average")
    plt.xlabel("Time")
    plt.ylabel("Price")
    plt.xticks(rotation=45)
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()
else:
    print(" Not enough data to plot.")
```

```python
###  Cell 10: BTC Price Line Plot

plt.figure(figsize=(12, 5))
plt.plot(df.index, df['price'], label='BTC Price', linewidth=2)
plt.title("Bitcoin Price Over Time", fontsize=14)
plt.xlabel("Time", fontsize=12)
plt.ylabel("Price (USD)", fontsize=12)
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
### Cell 11: Histogram of BTC Prices

plt.figure(figsize=(10, 4))
plt.hist(df['price'], bins=30, color='steelblue', edgecolor='black')
plt.axvline(df['price'].mean(), color='red', linestyle='--', label=f"Mean: ${df['price'].mean():,.0f}")
plt.axvline(df['price'].median(), color='green', linestyle='--', label=f"Median: ${df['price'].median():,.0f}")
plt.title("Distribution of Bitcoin Prices", fontsize=14)
plt.xlabel("Price (USD)", fontsize=12)
plt.ylabel("Frequency", fontsize=12)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
### Cell 12: Rolling Average and Volatility Band

df['rolling_avg_7'] = df['price'].rolling(window=7).mean()
df['rolling_std_7'] = df['price'].rolling(window=7).std()

plt.figure(figsize=(12, 5))
plt.plot(df.index, df['price'], label='Price', alpha=0.6)
plt.plot(df.index, df['rolling_avg_7'], label='7-period MA', linestyle='--', color='darkblue')
plt.fill_between(df.index,
                 df['rolling_avg_7'] - df['rolling_std_7'],
                 df['rolling_avg_7'] + df['rolling_std_7'],
                 color='orange', alpha=0.3, label='Volatility Band')
plt.title("📉 BTC Price with Rolling Average and Volatility", fontsize=14)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
