## Step 1: Import Required Modules
We begin by importing the functions defined in `utils.py`. These include:
- `init_db()` to create the database
- `get_session()` to start a session
- `fetch_price()` to get the real-time Bitcoin price
- `save_price()` to insert that price into our database


```python
from utils import init_db, get_session, fetch_30day_price_series, save_price_series, fetch_price, save_price

# Step 1: Set up DB
engine = init_db()
session = get_session(engine)

# Step 2: Fetch and store 30-day history
series = fetch_30day_price_series()
save_price_series(session, series)

# Step 3: Add current real-time price
current_price = fetch_price()
save_price(session, current_price)

from utils import fetch_realtime_5min_series, save_price_series

series = fetch_realtime_5min_series(interval_seconds=15)
save_price_series(session, series)

```
