## Step 1: Import Required Modules
We begin by importing the functions defined in `utils.py`. These include:
- `init_db()` to create the database
- `get_session()` to start a session
- `fetch_price()` to get the real-time Bitcoin price
- `save_price()` to insert that price into our database


```python
from utils import init_db, get_session, fetch_price, save_price
```

```python
engine = init_db()  # Creates bitcoin_data.db file
session = get_session(engine)
```

```python
price = fetch_price()
save_price(session, price)
```

```python
from utils import BitcoinPrice

records = session.query(BitcoinPrice).all()
for row in records:
    print(row.timestamp, row.price)
```
