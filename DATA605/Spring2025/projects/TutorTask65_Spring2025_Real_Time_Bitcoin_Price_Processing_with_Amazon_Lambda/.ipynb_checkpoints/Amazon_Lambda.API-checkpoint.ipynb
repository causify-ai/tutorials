# Amazon Lambda API – Real-Time BTC Price Fetch + Upload

This notebook simulates a Lambda function:
- Fetches Bitcoin price in real-time using the Coinbase API.
- Uploads the result as a JSON file into your AWS S3 bucket.


```python
from amazon_lambda_utils import get_live_btc_price, upload_to_s3

# AWS bucket and file settings
bucket_name = "ruthvick-btc-bucket"
file_name = "btc_price_latest.json"

# Step 1: Fetch Live BTC Price
print("Fetching live Bitcoin price...")
btc_price_data = get_live_btc_price()
print("Fetched data:", btc_price_data)

# Step 2: Upload to S3
print("\n Uploading to AWS S3...")
upload_result = upload_to_s3(btc_price_data, bucket_name, file_name)
print(f"Upload successful! File uploaded to: {upload_result}")
```
