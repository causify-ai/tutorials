# Real-Time Bitcoin Price Processing Example
This notebook demonstrates the full flow of simulating AWS Lambda functions to fetch live Bitcoin price data and upload it to AWS S3 storage.


## Step 1: Fetch Live Bitcoin Price Data
We simulate fetching real-time Bitcoin price using a WebSocket-like API (Coinbase spot price API).


```python
import datetime
from amazon_lambda_utils import get_live_btc_price, upload_to_s3
```

```python
# Fetch live Bitcoin price data
btc_price_data = get_live_btc_price()
print(f"Live Bitcoin Price Data:\n{btc_price_data}")
```

## Step 2: Upload Bitcoin Price Data to AWS S3
The fetched Bitcoin price is saved into an AWS S3 bucket in JSON format for storage and further analysis.


```python
# Define S3 bucket and filename
bucket_name = "ruthvick-btc-bucket"
timestamp = datetime.datetime.utcnow().strftime('%Y%m%dT%H%M%S')
file_name = f"btc_data/btc_price_{timestamp}.json"

# Upload the data
upload_result = upload_to_s3(btc_price_data, bucket_name, file_name)
print(f"Successfully uploaded to: {upload_result}")
```

# Final Summary
- Fetched real-time Bitcoin price from Coinbase API.
- Uploaded data into AWS S3 with a timestamped filename.
- This simulates how AWS Lambda functions can automate real-time data ingestion pipelines.
