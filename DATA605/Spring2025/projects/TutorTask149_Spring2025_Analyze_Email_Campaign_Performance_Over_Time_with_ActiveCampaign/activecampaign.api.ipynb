```python
import os
import requests
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
from dotenv import load_dotenv

%matplotlib inline
sns.set(style="whitegrid")
```

```python
load_dotenv()  # This loads variables from .env file

API_URL = os.getenv("ACTIVE_CAMPAIGN_API_URL")
API_KEY = os.getenv("ACTIVE_CAMPAIGN_API_KEY")

HEADERS = {
    "Api-Token": API_KEY,
    "Content-Type": "application/json"
}
```

```python
def get_campaigns() -> pd.DataFrame:
    """
    Fetch campaign metadata from the ActiveCampaign API.
    Returns a DataFrame with basic info like campaign ID, name, and date.
    """
    url = f"{API_URL}/api/3/campaigns"
    response = requests.get(url, headers=HEADERS)
    
    if response.status_code == 200:
        return pd.DataFrame(response.json().get("campaigns", []))
    else:
        raise Exception(f"API error: {response.status_code} - {response.text}")
```

```python
campaign_df = get_campaigns()
campaign_df.head()
```

```python
campaign_df.to_csv("activecampaign_api_data.csv", index=False)
print("Campaign metadata saved to 'activecampaign_api_data.csv'")
```

Now that we’ve collected the campaign metadata, we’ll use this CSV file (`activecampaign_api_data.csv`) in our main analysis notebook.
