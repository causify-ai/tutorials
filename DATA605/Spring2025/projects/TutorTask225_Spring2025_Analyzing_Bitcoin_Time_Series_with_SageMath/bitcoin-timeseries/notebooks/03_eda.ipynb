## 1. Load Preprocessed Data

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/bitcoin_timeseries.csv')
df['timestamp'] = pd.to_datetime(df['timestamp'])
df = df.sort_values('timestamp')
df.set_index('timestamp', inplace=True)
```

## 2. Plot Full BTC Time Series

```python
plt.figure(figsize=(15, 5))
df['price_usd'].plot(title='Bitcoin Price Over Time')
plt.xlabel("Timestamp")
plt.ylabel("Price (USD)")
plt.grid(True)
plt.tight_layout()
plt.savefig('../reports/full_timeseries.png')
plt.show()
```

## 3. Overlay Rolling Averages (24h & 7d)

```python
df['rolling_24h'] = df['price_usd'].rolling(window=24).mean()
df['rolling_7d'] = df['price_usd'].rolling(window=24*7).mean()

plt.figure(figsize=(15, 5))
df['price_usd'].plot(label='Price', alpha=0.5)
df['rolling_24h'].plot(label='24h Rolling Avg')
df['rolling_7d'].plot(label='7d Rolling Avg')
plt.legend()
plt.title("Bitcoin Price with Rolling Averages")
plt.grid(True)
plt.tight_layout()
plt.savefig('../reports/rolling_averages.png')
plt.show()
```

## 4. Price Distribution Histogram

```python
plt.figure(figsize=(10, 4))
df['price_usd'].hist(bins=100)
plt.title("Distribution of BTC Prices")
plt.xlabel("Price (USD)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig('../reports/histogram_prices.png')
plt.show()
```

## 5. Zoom into Recent Data (Last 30 & 60 Days)

```python
last_timestamp = df.index.max()

last_60 = df.loc[df.index >= last_timestamp - pd.Timedelta(days=60)]
last_30 = df.loc[df.index >= last_timestamp - pd.Timedelta(days=30)]

plt.figure(figsize=(15, 5))
last_60['price_usd'].plot(label='Last 60 Days', alpha=0.6)
last_30['price_usd'].plot(label='Last 30 Days', alpha=0.8)
plt.legend()
plt.title("BTC Price - Last 60 vs Last 30 Days")
plt.grid(True)
plt.tight_layout()
plt.savefig('../reports/zoomed_recent_60_30_days.png')
plt.show()
```

## 6. Volatility Estimate (Rolling Std Dev)

```python
df['volatility_24h'] = df['price_usd'].rolling(window=24).std()

plt.figure(figsize=(15, 5))
df['volatility_24h'].plot(title='24h Rolling Volatility')
plt.xlabel("Timestamp")
plt.ylabel("Volatility (Standard Deviation)")
plt.grid(True)
plt.tight_layout()
plt.savefig('../reports/volatility_rolling_stddev.png')
plt.show()
```
