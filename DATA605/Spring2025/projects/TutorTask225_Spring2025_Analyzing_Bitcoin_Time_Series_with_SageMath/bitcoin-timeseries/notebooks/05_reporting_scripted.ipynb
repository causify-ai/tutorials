## Final Combined Plot

```python
from PIL import Image
import matplotlib.pyplot as plt

# Load individual forecast images
symbolic_img = Image.open('../reports/symbolic_forecast_polydeg5.png')
arima_img = Image.open('../reports/arima_forecast.png')
sarima_img = Image.open('../reports/arima_vs_sarima_comparison.png')

# Create a horizontal composite plot
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

axes[0].imshow(symbolic_img)
axes[0].set_title('Symbolic Forecast')
axes[0].axis('off')

axes[1].imshow(arima_img)
axes[1].set_title('ARIMA Forecast')
axes[1].axis('off')

axes[2].imshow(sarima_img)
axes[2].set_title('SARIMA Forecast')
axes[2].axis('off')

plt.suptitle('Final Forecast Comparison: Symbolic vs ARIMA vs SARIMA', fontsize=16)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.savefig('../reports/final_plot.png')
plt.show()
```
