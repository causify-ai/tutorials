## 1. Load and Inspect Merged BTC Time Series

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load combined data
df = pd.read_csv('../data/bitcoin_timeseries.csv')
df['timestamp'] = pd.to_datetime(df['timestamp'])
df = df.sort_values('timestamp')

print(df.head())
print(f"✅ Loaded {len(df)} rows")
```

## 2. Check for Duplicates and Missing Timestamps

```python
# Drop duplicates if any
df = df.drop_duplicates(subset='timestamp')

# Optional: Set index for resampling
df.set_index('timestamp', inplace=True)

# Show gaps in time if any (e.g., using resample logic)
df = df[~df.index.duplicated()]
df = df.sort_index()

print(f"After cleanup: {len(df)} rows")
```

## 3. Optional: Resample to Hourly or Daily

```python
df_hourly = df.resample('1h').mean().dropna()
print(df_hourly.head())
```

## 4. Add Rolling Averages

```python
df_hourly['rolling_24h'] = df_hourly['price_usd'].rolling(window=24).mean()
df_hourly['rolling_7d'] = df_hourly['price_usd'].rolling(window=24*7).mean()
df_hourly.tail()
```
