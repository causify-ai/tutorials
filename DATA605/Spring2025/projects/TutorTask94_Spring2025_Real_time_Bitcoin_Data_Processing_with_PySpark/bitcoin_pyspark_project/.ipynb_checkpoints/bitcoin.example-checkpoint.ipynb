```python
from bitcoin_utils import get_spark_session
from pyspark.sql.functions import window, avg

spark = get_spark_session("BitcoinAnalysis")

# Read the data collected by the streaming job
df = spark.read.parquet("hdfs://hadoop:9000/data/bitcoin_prices")

# Ensure the data is sorted by time
df = df.orderBy("timestamp")
df.show(5)
```

```python
# Calculate a moving average of the price over a 2-point window (i.e., averaging current and previous price)
from pyspark.sql.window import Window
import pyspark.sql.functions as F

# Define a window spec that includes the current row and 1 previous row, ordered by timestamp
winSpec = Window.orderBy("timestamp").rowsBetween(-1, 0)
df_with_ma = df.withColumn("moving_avg_price", F.avg("current_price").over(winSpec))
df_with_ma.show()
```

```python
# Example of time-based window average (e.g., average price per 1 minute window)
df_time_avg = df.groupBy(window("timestamp", "1 minutes")).agg(avg("current_price").alias("avg_price"))
df_time_avg.show()
```

```python
# Convert to Pandas for visualization
pdf = df_with_ma.toPandas()
print(pdf.head())
```

```python
import matplotlib.pyplot as plt

# Ensure timestamp is datetime type in pandas
pdf['timestamp'] = pandas.to_datetime(pdf['timestamp'])

plt.figure(figsize=(8,4))
plt.plot(pdf['timestamp'], pdf['current_price'], marker='o', label='Price')
plt.plot(pdf['timestamp'], pdf['moving_avg_price'], marker='x', label='2-pt Moving Avg')
plt.title("Bitcoin Price Over Time")
plt.xlabel("Time")
plt.ylabel("Price (USD)")
plt.legend()
plt.grid(True)
plt.show()
```
