```python
from bitcoin_utils import get_spark_session, stream_prices_to_hdfs

# Create Spark session (or get existing one)
spark = get_spark_session("BitcoinPriceStreamingApp")

# Define HDFS output path for streaming data
# The HDFS namenode is accessible via the hostname 'hadoop' (from docker-compose).
output_path = "hdfs://hadoop:9000/data/bitcoin_prices" 
```

```python
# Stream data for 2 minutes, fetching every 30 seconds (this will fetch ~4 times)
stream_prices_to_hdfs(spark, output_dir=output_path, interval_sec=30, run_minutes=2)
```

```python
# Read back the data from HDFS to ensure it’s there
df = spark.read.parquet(output_path)
df.show()
```
