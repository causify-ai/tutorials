```python
import requests
import pandas as pd
import time

while True:
    try:
        # Step 1: API call to get the latest 1-minute OHLC data
        url = 'https://api.binance.com/api/v3/klines'
        params = {
            'symbol': 'BTCUSDT',
            'interval': '1m',
            'limit': 1
        }
        
        response = requests.get(url, params=params)
        response.raise_for_status()  # Raise an error if request failed
        data = response.json()[0]    # Get the latest 1-minute candle

        # Step 2: Parse the data
        candle = {
            'Open Time': pd.to_datetime(data[0], unit='ms'),
            'Open': float(data[1]),
            'High': float(data[2]),
            'Low': float(data[3]),
            'Close': float(data[4]),
            'Volume': float(data[5]),
            'Number of Trades': int(data[8])
        }

        # Step 3: Create a DataFrame
        df = pd.DataFrame([candle])

        # Step 4: Show the DataFrame
        print(df)

        # Optional: Save to CSV or database if needed
        # df.to_csv('btc_prices.csv', mode='a', header=False, index=False)

    except Exception as e:
        print(f"Error fetching data: {e}")

    # Step 5: Wait for 1 minute
    time.sleep(60)
```
