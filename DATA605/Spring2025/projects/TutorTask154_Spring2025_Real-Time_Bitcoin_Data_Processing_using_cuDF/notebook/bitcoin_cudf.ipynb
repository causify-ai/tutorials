# Real-Time Bitcoin Data Processing using cuDF
This notebook demonstrates real-time data fetching, storage, and visualization using CoinGecko API and cuDF.

```python
import sys, os
sys.path.append(os.path.abspath("../utils"))

from fetch_price import get_bitcoin_price
from process_data import add_to_dataframe, compute_moving_averages, compute_volatility, compute_rate_of_change, save_to_csv
from visualize import plot_all

import cudf
import time
```

## Initialize DataFrame

```python
gdf = cudf.DataFrame(columns=["timestamp", "price"])
```

## Collect Live Bitcoin Prices

```python
for _ in range(100):
    ts, price = get_bitcoin_price()
    gdf = add_to_dataframe(gdf, ts, price)
    print(f"Fetched @ {ts} --> ${price}")
    time.sleep(20)
```

## Time series analysis

```python
gdf = compute_moving_averages(gdf, window=5)
gdf = compute_volatility(gdf, window=5)
gdf = compute_rate_of_change(gdf, periods=1)

# Save to CSV
save_to_csv(gdf, filename="bitcoin_prices.csv")

print("Saved Bitcoin prices with analysis to bitcoin_prices.csv")
```

## Visualize results

```python
plot_all(gdf)
```
