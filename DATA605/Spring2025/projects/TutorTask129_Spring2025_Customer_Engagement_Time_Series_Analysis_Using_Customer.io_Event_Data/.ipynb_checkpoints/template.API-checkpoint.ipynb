## Create user on customer.io

```python
from customerio import CustomerIO
from faker import Faker
import pandas as pd
import time
import random
from datetime import datetime, timedelta

# === SETUP ===
SITE_ID = "1d7fdb608de0a0a8cd66"
API_KEY = "4c9b87c8eae4e9729835"
cio = CustomerIO(site_id=SITE_ID, api_key=API_KEY)
faker = Faker()

# === GENERATE USERS ===
num_users = 1000
user_ids = []

for i in range(num_users):
    user_id = f"user_{faker.uuid4()[:8]}"
    email = faker.email()
    name = faker.name()

    cio.identify(id=user_id, email=email, name=name)
    user_ids.append(user_id)
    time.sleep(0.05)

pd.DataFrame({"user_id": user_ids}).to_csv("simulated_user_ids.csv", index=False)
print("✅ Created and saved 1000 users.")
```

## Simulate the user event

```python
# === SIMULATE EVENTS ===
event_types = ["email_opened", "clicked", "app_login"]
campaigns = ["Spring Sale", "Black Friday", "Summer Promo"]
days_back = 180
all_events = []

for user_id in user_ids:
    for _ in range(random.randint(30, 60)):
        event_type = random.choice(event_types)
        campaign = random.choice(campaigns)
        days_ago = random.randint(0, days_back)
        timestamp = int((datetime.now() - timedelta(days=days_ago)).timestamp())
        device = random.choice(["iPhone", "Android", "Web"])

        cio.track(
            customer_id=user_id,
            name=event_type,
            data={"campaign": campaign, "device": device},
            timestamp=timestamp
        )

        all_events.append({
            "user_id": user_id,
            "event_name": event_type,
            "timestamp": pd.to_datetime(timestamp, unit="s"),
            "campaign": campaign,
            "device": device
        })

    time.sleep(0.2)

pd.DataFrame(all_events).to_csv("simulated_event_log.csv", index=False)
print("Simulated events saved to 'simulated_event_log.csv'")
```
