## Load Transformers Pipelines

```python
from transformers import pipeline

ner_pipeline = pipeline("ner", model="dslim/bert-base-NER", aggregation_strategy="simple")
classifier_pipeline = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")
```

## Example: Named Entity Recognition (NER)

```python
sample_text = "Bitcoin price surged after SEC approved the ETF proposal for Blackrock."
ner_results = ner_pipeline(sample_text[:512])
for ent in ner_results:
    print(ent)
```

## Example: Zero-Shot Event Classification

```python
candidate_labels = [
    "Regulatory", "Technological", "Market Manipulation", "Adoption News",
    "ETF News", "Security Breach", "Legal Action", "Macro News"
]

classification_result = classifier_pipeline(sample_text[:512], candidate_labels)
print(f"Predicted event type: {classification_result['labels'][0]}")
```
