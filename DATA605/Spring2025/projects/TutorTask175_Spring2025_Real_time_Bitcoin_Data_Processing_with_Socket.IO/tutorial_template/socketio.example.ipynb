```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from socketio_utils import compute_sma
import random
```

```python
# Generate simulated Bitcoin prices
def generate_fake_prices(n=30, base=27500):
    prices = []
    for _ in range(n):
        change = random.uniform(-50, 50)
        base += change
        prices.append(round(base, 2))
    return prices

btc_prices = generate_fake_prices()
btc_sma = [None if i < 4 else compute_sma(btc_prices[:i+1], window=5) for i in range(len(btc_prices))]

df = pd.DataFrame({'Price': btc_prices, 'SMA_5': btc_sma})
df.head()
```

This chart shows a simulated Bitcoin price stream with a 5-point Simple Moving Average (SMA) overlay. 

```python
# Create a Plotly chart of price + SMA
fig = go.Figure()
fig.add_trace(go.Scatter(y=df['Price'], mode='lines+markers', name='BTC Price'))
fig.add_trace(go.Scatter(y=df['SMA_5'], mode='lines', name='5-pt SMA'))

fig.update_layout(
    title='Simulated Bitcoin Price with SMA',
    xaxis_title='Time Step',
    yaxis_title='Price (USD)',
    template='plotly_dark',
    width=900,
    height=500
)
fig.show()
```

The price values are generated using small random fluctuations to simulate a live feed. The SMA helps visualize short-term trends and smooths out volatility.
