This notebook shows a simple example of how to use the Bitcoin Trade Processing API.

- We process it using the API functions.
- Monitoring, aggregation, anomaly detection, sentiment analysis, and Slack alerts happen automatically.


```python
# Import the main API function
from btc_trade_API import start_monitoring, process_trade
```
