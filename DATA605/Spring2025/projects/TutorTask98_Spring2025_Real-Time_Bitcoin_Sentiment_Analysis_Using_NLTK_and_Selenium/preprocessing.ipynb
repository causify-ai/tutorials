```python
import pandas as pd
import re
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import os
```

```python
# Download NLTK resources if not already present
nltk.download('punkt')
nltk.download('stopwords')
```

```python
df = pd.read_csv("tweets/all_tweets.csv")
```

```python
def clean_text(text):
    # Remove URLs, mentions, hashtags, special characters, and digits
    text = re.sub(r"http\S+|www\S+|https\S+", '', text, flags=re.MULTILINE)  # URLs
    text = re.sub(r'\@\w+|\#', '', text)  # mentions and hashtags
    text = re.sub(r"[^a-zA-Z\s]", '', text)  # special characters and numbers
    text = text.lower()  # lowercase
    return text

# Tokenization and stopword removal
stop_words = set(stopwords.words('english'))

def tokenize_and_remove_stopwords(text):
    tokens = word_tokenize(text)
    filtered = [word for word in tokens if word not in stop_words and len(word) > 1]
    return filtered
```

```python
# Apply preprocessing steps
df['cleaned_text'] = df['Content'].astype(str).apply(clean_text)
df['tokens'] = df['cleaned_text'].apply(tokenize_and_remove_stopwords)
```

```python
# Save to a new CSV
PREPROCESSED_PATH = os.path.join('preprocessed_tweets.csv')

def save_preprocessed_to_csv(new_df, file_path=PREPROCESSED_PATH):
    # Check if preprocessed CSV already exists
    if os.path.exists(file_path):
        existing_df = pd.read_csv(file_path)
        combined_df = pd.concat([existing_df, new_df], ignore_index=True)
    else:
        combined_df = new_df
    
    # Drop duplicates based on 'Tweet ID'
    if 'Tweet ID' in combined_df.columns:
        combined_df = combined_df.drop_duplicates(subset=['Tweet ID'], keep='last')

    combined_df.to_csv(file_path, index=False, encoding="utf-8")
    print(f"[INFO] Preprocessed data saved to: {file_path} | Total rows: {len(combined_df)}")

# Usage after your preprocessing
save_preprocessed_to_csv(df)
```
