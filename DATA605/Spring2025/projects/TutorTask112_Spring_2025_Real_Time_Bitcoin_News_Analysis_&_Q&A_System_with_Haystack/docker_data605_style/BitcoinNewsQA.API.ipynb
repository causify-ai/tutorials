```python
# BitcoinNewsQA.API.ipynb

# Step 1: Import Utility Functions
from BitcoinNewsQA_utils import fetch_crypto_news, create_documents, split_documents

# Step 2: Fetch Raw News from CryptoPanic
news_data = fetch_crypto_news()
print(f"Fetched {len(news_data)} articles.")
news_data[0]
```

```python
# Step 3: Convert Articles to Haystack Documents
documents = create_documents(news_data)
print(f"Converted to {len(documents)} Haystack documents.")
documents[0].to_dict()
```
