```python
import requests
import datetime

# CryptoPanic API key
API_KEY = "f2b5f828c73705d7c2f8a681e5dbc53a3357af38"

def fetch_bitcoin_news():
    url = f"https://cryptopanic.com/api/v1/posts/?auth_token={API_KEY}&currencies=BTC"
    response = requests.get(url)

    if response.status_code != 200:
        raise Exception(f"Failed to fetch news: {response.status_code} | {response.text}")

    data = response.json()
    articles = []

    for item in data.get("results", []):
        articles.append({
            "title": item.get("title", ""),
            "published_at": item.get("published_at", ""),
            "url": item.get("url", ""),
            "text": item.get("body") or item.get("title", ""),
            "source": item.get("source", {}).get("title", "")
        })

    return articles

# Fetch and preview
bitcoin_articles = fetch_bitcoin_news()
print(f" Fetched {len(bitcoin_articles)} articles.")
print(" Sample article:\n", bitcoin_articles[0])
```
