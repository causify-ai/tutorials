# bitcoin_emr.API.ipynb

This notebook demonstrates the core utility functions for fetching and storing Bitcoin price data. These functions are defined in the `bitcoin_emr_utils.py` file and form the foundation of the real-time data ingestion pipeline.


```python
import sys
sys.path.append('bitcoin_kafka')

from bitcoin_emr_utils import fetch_bitcoin_price, get_current_timestamp, save_price_to_s3
```

## Fetch Current Bitcoin Price

We use the `fetch_bitcoin_price()` function to retrieve the latest Bitcoin price from the CoinGecko API.


```python
price = fetch_bitcoin_price()
print("Fetched Bitcoin Price (USD):", price)
```

## Generate Current Timestamp

We use the `get_current_timestamp()` utility to record the exact time when the price is fetched.


```python
timestamp = get_current_timestamp()
print("Current Timestamp:", timestamp)
```

## Save Record to Amazon S3

This function combines the price and timestamp into a JSON object and uploads it to the specified S3 bucket.


```python
save_price_to_s3(bucket='bitcoin-price-streaming-data', folder='data_v2', price_usd=price)
```
