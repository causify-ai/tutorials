# bitcoin_emr.example.ipynb

This notebook simulates a real-time ingestion loop that fetches Bitcoin price data from an API and stores it to Amazon S3. This mimics how a producer script would operate in a live streaming pipeline.


```python
import sys
sys.path.append('bitcoin_kafka')

from bitcoin_emr_utils import fetch_bitcoin_price, save_price_to_s3
import time
```

## Simulate Real-Time Ingestion

We'll fetch Bitcoin prices and upload them to S3 in a short loop with a 5-second delay between iterations.


```python
for i in range(3):  # simulate 3 records
    price = fetch_bitcoin_price()
    save_price_to_s3(bucket="bitcoin-price-streaming-data", folder="data_v2", price_usd=price)
    print(f"Record {i+1} uploaded.\n")
    time.sleep(60)
```

## Summary

This notebook successfully simulates a real-time data ingestion pipeline. Each iteration fetches a fresh Bitcoin price and stores it in S3. This demonstrates the functionality of the utility layer and mirrors the behavior of `bitcoin_producer.py`.
