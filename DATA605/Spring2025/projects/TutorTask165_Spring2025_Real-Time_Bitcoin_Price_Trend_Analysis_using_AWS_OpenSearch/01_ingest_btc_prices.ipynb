```python
# 1. Setup

import sys
sys.path.append("..") 

import os
print(os.listdir("../utils"))

from pycoingecko import CoinGeckoAPI
from datetime import datetime
from utils.opensearch_client import get_opensearch_client
import pandas as pd

cg = CoinGeckoAPI()
client = get_opensearch_client("search-bitcoin-price-analysis-amarxtyvowvlm6ufg4apvax5ru.us-east-2.es.amazonaws.com")


# 2. Create Index (once)
index_name = "btc-prices"
client.indices.create(index=index_name, ignore=400)

# 3. Fetch Latest Price
price_data = cg.get_price(ids="bitcoin", vs_currencies="usd")
timestamp = datetime.utcnow().isoformat()
doc = {
    "timestamp": timestamp,
    "price": price_data["bitcoin"]["usd"]
}

# 4. Index Document
client.index(index=index_name, body=doc)

# 5. (Optional) Save Locally
df = pd.DataFrame([doc])
df.to_csv("../data/btc_price_history.csv", mode='a', header=False, index=False)
```
