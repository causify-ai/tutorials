```python
# Import utilities
from btc_utils import (
    get_btc_history,
    get_gold_price,
    calculate_volatility,
    detect_anomalies,
    calculate_correlation,
    forecast_btc_prices
)

import plotly.express as px
import plotly.graph_objects as go
```

```python
# Load historical BTC and gold prices
btc_df = get_btc_history(days=90)       # Returns DataFrame with 'date' and 'price'
gold_df = get_gold_price()              # Returns DataFrame with 'date' and 'gold_price'
```

```python
btc_df = calculate_volatility(btc_df)  # Adds 'volatility_30d' column

# Plot price and volatility
fig = go.Figure()
fig.add_trace(go.Scatter(x=btc_df["date"], y=btc_df["price"], name="BTC Price"))
fig.add_trace(go.Scatter(x=btc_df["date"], y=btc_df["volatility_30d"], name="30-Day Volatility", yaxis="y2"))

fig.update_layout(
    title="Bitcoin Price & 30-Day Volatility",
    yaxis=dict(title="Price"),
    yaxis2=dict(title="Volatility", overlaying="y", side="right")
)
fig.show()
```

```python
anomalies = detect_anomalies(btc_df)

# Show the anomalies
anomalies[["date", "price", "pct_change"]]
```

```python
merged_df = calculate_correlation(btc_df, gold_df)

# Plot correlation
fig = px.line(merged_df, x="date", y="rolling_corr_30d", title="Rolling Correlation (BTC vs Gold)")
fig.update_layout(yaxis=dict(range=[-1, 1]))
fig.show()
```

```python
forecast_days = 30
forecast_df, model = forecast_btc_prices(btc_df, periods=forecast_days)

from prophet.plot import plot_plotly
fig = plot_plotly(model, forecast_df)
fig.update_layout(title=f"Bitcoin Forecast for {forecast_days} Days")
fig.show()
```

This notebook demonstrates a complete workflow for real-time Bitcoin market analysis:
- Price ingestion
- Volatility and anomaly detection
- Correlation with gold
- Forecasting future prices

All functionality is modular and handled by the `btc_utils.py` utility file.
